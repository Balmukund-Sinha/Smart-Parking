# 02 - PySpark EDA

Exploratory analysis of the raw parking events using PySpark.
Requires PySpark installed locally (`pip install pyspark`).

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, avg, count

spark = SparkSession.builder.appName('ParkingEDA-Notebook').getOrCreate()
df = spark.read.option('header', True).option('inferSchema', True).csv('../data/raw/parking_events.csv')
df.printSchema()
```

```python
df.count()
```

## Zone-wise average occupancy

```python
df.groupBy('zone_id').agg(avg('occupancy').alias('avg_occupancy'), count('*').alias('events')).orderBy(col('avg_occupancy').desc()).show()
```

## Peak hours

```python
from pyspark.sql.functions import hour, to_timestamp

df2 = df.withColumn('timestamp', to_timestamp('timestamp')).withColumn('hour', hour('timestamp'))
df2.groupBy('hour').agg(avg('occupancy').alias('avg_occupancy')).orderBy(col('avg_occupancy').desc()).show(24)
```
