# 01 - Data Generation

Generates the synthetic parking-events dataset used by the rest of this project.
This wraps `scripts/generate_dataset.py` so you can explore the data interactively.

````python
import sys
sys.path.insert(0, '..')
from scripts.generate_dataset import build_zones, build_events, build_ml_dataset, build_dashboard_snapshots
import numpy as np
import pandas as pd
````

## Configure and generate

````python
rng = np.random.default_rng(42)
start_date = pd.Timestamp('2026-01-01')
num_days = 21

zones_df = build_zones(rng, num_zones=20, cap_min=30, cap_max=150, util_min=0.5, util_max=0.85, dur_min=45, dur_max=150)
zones_df.head()
````

````python
vehicle_ids = np.array([f'V{i+1:05d}' for i in range(6000)])
holidays = pd.to_datetime(['2026-01-01','2026-01-26']).date

events, w_temp, w_rain = build_events(rng, zones_df, vehicle_ids, num_days, start_date, holidays)
print(f'{len(events):,} events generated')
events.head()
````

## Quick sanity checks

````python
util = events['occupancy'] / events['capacity'] * 100
print('mean util %:', util.mean())
print('max util %:', util.max())
print('any occupancy > capacity:', (events.occupancy > events.capacity).sum())
````
